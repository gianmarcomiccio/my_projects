```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB
from sklearn.preprocessing import KBinsDiscretizer
from sklearn.metrics import accuracy_score, classification_report

# Load Iris dataset
iris = load_iris()
X = iris.data
y = iris.target

# Discretize continuous features into bins
# 'ordinal' encoding is required for MultinomialNB
discretizer = KBinsDiscretizer(n_bins=5, encode='ordinal', strategy='uniform')
X_binned = discretizer.fit_transform(X)

# Split into train and test sets
X_train, X_test, y_train, y_test = train_test_split(X_binned, y, test_size=0.3, random_state=42)

# Train Multinomial Naive Bayes
nb = MultinomialNB()
nb.fit(X_train, y_train)

# Predict
y_pred = nb.predict(X_test)

# Evaluate
print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred, target_names=iris.target_names))

```

```python
from sklearn.datasets import load_iris
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
import numpy as np

# Load Iris dataset
iris = load_iris()
X = iris.data
y = iris.target
feature_names = iris.feature_names
class_names = iris.target_names

# Convert to pandas DataFrame for easy viewing
df = pd.DataFrame(X, columns=feature_names)
df['species'] = [class_names[i] for i in y]

# Print first 5 rows
print("Head of the dataset:")
print(df.head())

# Split into train/test
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Train Gaussian Naive Bayes
gnb = GaussianNB()
gnb.fit(X_train, y_train)

# Print priors
print("\nClass prior probabilities:")
for name, prior in zip(class_names, gnb.class_prior_):
    print(f"{name}: {prior:.2f}")

# Print means and variances per class
print("\nFeature means per class:")
means_df = pd.DataFrame(gnb.theta_, columns=feature_names, index=class_names)
print(means_df)

print("\nFeature variances per class:")
vars_df = pd.DataFrame(gnb.sigma_, columns=feature_names, index=class_names)
print(vars_df)

# Example: compute posterior for first test sample
sample = X_test[0].reshape(1, -1)
posterior_probs = gnb.predict_proba(sample)
print("\nExample sample:", sample)
print("Posterior probabilities for each class:")
for name, prob in zip(class_names, posterior_probs[0]):
    print(f"{name}: {prob:.4f}")
```
